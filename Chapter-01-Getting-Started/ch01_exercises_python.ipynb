# Exercise 1 — Possible predictor variables for Cases 3 and 4

*For Cases 3 and 4 in Section 1.5, list possible predictor variables that might be useful,
assuming the relevant data are available.*

## Case 3 — resale value of leased vehicles

| Group | Candidate predictors |
|---|---|
| The vehicle | make, model, body type, engine size, fuel type, transmission, colour, options/trim level |
| Its use | odometer reading at sale, age at sale (lease length), accident and service history, condition grade |
| The lease | lessee type (private vs. corporate fleet), region where it was used |

| The sale | sales channel (auction vs. dealer), location of sale |

## Case 4 — weekly air passengers by route and class

| Group | Candidate predictors |
|---|---|
| Calendar | school holidays in each city (they differ by state), public holidays, Easter/Christmas, week of year |
| Events | major sporting events, conferences and festivals (and which city hosts them), strikes |
| Price & product | own ticket prices by class, competitors' prices, seat capacity per class (the reclassification trial), number of flights / aircraft size |
| Competition | entry/exit of competing (cut-price) airlines, competitor capacity |
| Marketing | advertising campaigns, loyalty-programme promotions |
| Economy | GDP / business activity (business class), consumer confidence, fuel prices, exchange rate (tourism) |

### Supporting analysis: what the `melsyd` data show

The weekly Ansett Melbourne–Sydney series shows why several of these predictors matter:
the 1989 pilots' strike (zero traffic), the 1992 seat reclassification (economy falls while
business rises), missing values, and a strong holiday effect at the start of each year.

```python
melsyd = fpp.load("melsyd")
econ = melsyd["Economy.Class"]
econ.attrs = dict(melsyd.attrs)
fpp.autoplot(econ, title="Economy class passengers: Melbourne–Sydney", xlab="Year", ylab="Thousands")
plt.show()

yearly = pd.DataFrame({"year": np.floor(fpp.ts_time(econ)).astype(int), "econ": econ.to_numpy()})
print(yearly.groupby("year")["econ"].mean().round(1))
print({"missing_weeks": int(econ.isna().sum()), "zero_weeks": int((econ == 0).sum())})
```
